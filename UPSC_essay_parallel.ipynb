```python
from langgraph.graph import StateGraph,START,END
from typing import TypedDict
from pydantic import BaseModel , Field
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI
```

```python
load_dotenv()
```

```python
# create llm
llm = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash",
    temperature=0
)
```

```python
#lets create schema for structured output did the same thing in langchain
class FeedbackAndNumber(BaseModel):
    text_feedback:str = Field(description=" A detailed feedback of essay")
    score:int = Field(description="score out of 10",ge=0,le=10)
```

```python
structured_llm = llm.with_structured_output(FeedbackAndNumber)
```
