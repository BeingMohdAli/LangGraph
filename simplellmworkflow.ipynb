```python
from langgraph.graph import StateGraph,START,END
from langchain_google_genai import ChatGoogleGenerativeAI
from typing import TypedDict
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
#first create state
class LLMQAState(TypedDict):
    question : str
    answer : str
```

```python
# create llm
llm = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash",
    temperature=0
)
```

```python
def qa(state:LLMQAState)->LLMQAState:
    question = state["question"]
    answer = llm.invoke(question)
    state["answer"] = answer.content
    return state
```

```python
#lets create graph
graph = StateGraph(LLMQAState)

#lets create node
graph.add_node("qa",qa)

#create edge
graph.add_edge(START,"qa")

graph.add_edge("qa",END)
```

```python
workflow = graph.compile()
```

```python
result = workflow.invoke({"question":"but how jmama masjid ws made in the way there is no technology"})
```

```python
print(result["answer"])
```
